```python
import chromadb
```

```python
# create chromadb client to connect with chroma database

client = chromadb.Client()
```

```python
# create collection to store the data
# collections are just like tables in RDBMS

collection1 = client.create_collection( name = "GenAI_AAA")
```

```python
# adding data in the collection
# add() function is used to insert the data into collection
collection1.add(
    documents = [
        "Python is a very simple programming language",
        "MYSQL is a RDBMS",
        "MAchine learning is a branch of AI",
        "Deep learning is based upon Neural networks"
    ],
    ids = [ "doc1", "doc2", "doc3", "doc4"]
)
```

```python
# get() function is used to get the data
result = collection1.get( ids = ["doc1", "doc2"])
result
```

```python
# update() function is used to update the documents

collection1.update(
    ids = ["doc1"],
    documents = ["Python is most popular language"])
```

```python
result = collection1.get( ids = ["doc1"])
result
```

```python
# delete() function to delete the document from cromedb

collection1.delete( ids = ["doc1"] )
```

```python
collection1.get()
```

```python
# Query() : to find the related chunks or docs inside the chromadb

results = collection1.query( query_texts = ["popular AI techniques"], n_results = 2)
results
```

# Metadata : data about data

```python
# Meta data make the searching of document easy
collection2 = client.create_collection( name = "GenAI_AAA2")
```

```python
collection2.add(
    documents = [
        "Python is a very simple programming language",
        "MYSQL is a RDBMS",
        "MAchine learning is a branch of AI",
        "Deep learning is based upon Neural networks"
    ],
    ids = [ "doc1", "doc2", "doc3", "doc4"],
    metadatas = [
        {"category" : "language"},
         {"category" : "RDBMS"},
         {"category" : "AI"},
         {"category" : "AI"}
    ]
        
)
```

```python
results = collection2.query( query_texts = ["popular AI techniques"], 
                                            n_results = 2, 
                                            where = {"category" : "AI"})
results
```

# Using ChromaDB with Sentence Transformer

```python
# using specific Embedding model for building Chromadb vector database

!pip install sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer
```

```python
# picking the embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
texts = [
    "Python is most popular programming language",
    "Machine learning is a branch of AI"
]

embeddings =  embedding_model.encode(texts)
print(len(embeddings[0]))

# sentenance transformer embedding model convert the document into vector of size 384
```

```python
import chromadb
client = chromadb.Client()
```

```python
collection4 = client.get_or_create_collection( name = "documents")
```

```python
collection4.add(
    documents = [
        "Python is most popular programming language",
    "Machine learning is a branch of AI"
    ],
    embeddings = embeddings,  # we are inserting ST based embedding
    ids = ["doc1", "doc2"]
)
```

```python
result = collection4.query(
                            query_texts = [" AI is the future"],
                            n_results = 1
)

result
```

# ChromaDB with Langchain

```python
pip install langchain-chroma
```

```python
pip install langchain-huggingface
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
# embedding model
embeddings = HuggingFaceEmbeddings( 
                        model_name = "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vectorstore = Chroma(
                    collection_name = "my_documents",
                    embedding_function = embeddings,
                    persist_directory = "./Lang_Chroma_db"
)
```

```python
from langchain_core.documents import Document

documents = [
    Document(
        page_content = "Python is most popular programming language",
        metadata = {"topic" : "Python"}
    ),
    Document(
        page_content = "Machine learning is a branch of AI",
        metadata = {"topic" : "ML"}
    )
]
```

```python
vectorstore.add_documents(documents)
```

```python
result = vectorstore.similarity_search("AI is growing very fast", k = 1)
```

```python
result
```

# Chromadb with ollama embedding model and ollama LLM

```python
import ollama
import chromadb
```

```python
client = chromadb.PersistentClient( path = "./ollama_chromdb")
```

```python
collection5 = client.get_or_create_collection( name = "documents")
```

```python
documents = [
        "Python is a very simple programming language",
        "MYSQL is a RDBMS",
        "MAchine learning is a branch of AI",
        "Deep learning is based upon Neural networks"
    ]

ids = [ "doc1", "doc2", "doc3", "doc4"]
```

```python
!ollama pull nomic-embed-text
```

```python
# using ollama based embedding model 
embeddings = []

for d in documents:
    response = ollama.embed( model = "nomic-embed-text", input = d)
    embeddings.append(response["embeddings"][0])
```

```python
collection5.add(
    documents = documents,
    ids = ids,
    embeddings = embeddings
)
```

```python
Query = "AI is powerful tool"

response = ollama.embed(model = "nomic-embed-text", input = Query)

query_embedding = response["embeddings"][0]
```

```python
results = collection5.query( query_embeddings = [query_embedding], n_results = 2)
results
```
