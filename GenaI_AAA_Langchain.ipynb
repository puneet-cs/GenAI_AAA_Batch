```python
# PromptTemplate : used to customize the prompt, and for making then resusable
# e.g : write a poem on {topic}

from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("write a poem on {topic}")

llm_obj = ChatGroq(
                    model = "openai/gpt-oss-20b",
                    temperature = 0.2
)

chain = prompt | llm_obj

user_input = input("Enter a topic")

response = chain.invoke({
                            "topic" : user_input
})

print(response)
```

```python
# format the output 
# StrOutputParser : used for getting output in proper string format

from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

llm_obj = ChatGroq(
                    model = "openai/gpt-oss-20b",
                    temperature = 0.2
)

chain = prompt | llm_obj | parser

user_input = input("Enter a topic")

response = chain.invoke({
                            "topic" : user_input
})

print(response)
```

```python
# getting output from LLM in JSON format using Langchain
# raising exception in langchain

# client -> Langchain -> LLM

from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("provide 10 json type data items on {topic}")
```

```python
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.exceptions import OutputParserException
```

```python
from langchain_groq import ChatGroq

parser = JsonOutputParser()

llm_obj = ChatGroq(model_name = "openai/gpt-oss-20b")

chain = prompt | llm_obj | parser

user_topic = input("Enter your topic: ")

try:
    response = chain.invoke({"topic" : user_topic})
    print(response)
except OutputParserException as e:
    print(f"Output is not in json format : {e}")
```
